```python
# 1. Payment Type Dimension
df_payment = spark.sql("""
    SELECT * FROM VALUES
    (-1, 'Undefined'),
    (1, 'Credit Card'),
    (2, 'Cash'),
    (3, 'No Charge'),
    (4, 'Dispute'),
    (5, 'Unknown'),
    (6, 'Voided Trip')
    AS payment_type(payment_type, payment_desc)
""")
df_payment.write.format("delta").mode("overwrite").saveAsTable("silvercleansed.dim_payment_type")

```

```python
# 2. Trip Type Dimension
df_trip = spark.sql("""
    SELECT * FROM VALUES
    (-1, 'Undefined'),
    (1, 'Street-hail'),
    (2, 'Dispatch')
    AS trip_type(trip_type, trip_desc)
""")
df_trip.write.format("delta").mode("overwrite").saveAsTable("silvercleansed.dim_trip_type")

```

```python
# 3. Vendor Dimension
df_vendor = spark.sql("""
    SELECT * FROM VALUES
    (-1, 'Undefined'),
    (1, 'Creative Mobile Technologies'),
    (2, 'Verifone Inc.')
    AS vendor_id(vendor_id, vendor_name)
""")
df_vendor.write.format("delta").mode("overwrite").saveAsTable("silvercleansed.dim_vendor")

```
